```python
from dotenv import load_dotenv

load_dotenv("./.env")
```

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
```

```python
model = "llama3.2"
base_url = "http://localhost:11434"
llm = ChatOllama(model=model, base_url=base_url, temperature=0.8)
```

```python
question_template = ChatPromptTemplate.from_template("""
                                                     You are a professor of AI computer science.
                                                     Please ask a beginning-level question about {topic} to a student.
                                                     Your question should be simple and easy to understand.
                                                     Your question:
                                                     """)
answer_template = ChatPromptTemplate.from_template("""
                                                   You are a student of AI computer science.
                                                   Please answer the professor's question in short points about AI.
                                                   Question: {question}
                                                   Your answer:
                                                   """)
```

```python
from langchain_core.output_parsers import StrOutputParser
question_chain = question_template | llm | StrOutputParser()
answer_chain = answer_template | llm | StrOutputParser()
```

```python
from langchain_core.runnables import RunnablePassthrough, RunnableParallel
```

```python
whole_chain = {"question": question_chain} | RunnableParallel({"question": RunnablePassthrough(), "answer":answer_chain} )
```

```python
topic = "Artificial Intelligence"
response = whole_chain.invoke({"topic":topic})
print("Question:", response["question"])
print("Answer:", response["answer"])    
```
