```python
from dotenv import load_dotenv
import os

load_dotenv("./../.env")
langsmith_api_key = os.getenv("LANGSMITH_API_KEY")
if langsmith_api_key:
    print("LangSmith API key loaded successfully.")
```

```python
from langchain_ollama import ChatOllama
base_url = "http://localhost:11434"
model = "llama3.2"

AI = ChatOllama(model=model, base_url=base_url, temperature=1, num_predict=256)
response = AI.invoke("Hello!")
print(response.content)
```

```python
for chunk in AI.stream("Hello, how are you?"):
    print(chunk.content, end="\n")
```

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
question_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant, you answer question like a {role}."),
    ("human", "Tell me {points} facts about {topics}."),
])
question = question_template.invoke({"role": "Cowboy", "points": 3, "topics": "the history of horses"})
print(question)
```

```python
response = AI.invoke(question)
print(response.content)
```
